```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/Applications/Repos/Repo/expedition/Python/Stocks/World-Stock-Prices-Dataset.csv')
```

```python
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'],utc=True).dt.date        # Without .dt, Pandas wouldn’t know you want to pick apart the datetime into pieces.
```

```python
df.head()
```

```python
df = df.round(2)            # rounds off every numerical value 
```

```python
df.head()
```

```python
df = df[df['Country'].isin(['usa'])]
```

```python
df.head()
```

Daily Percent Change

```python
df['Daily_Return_%'] = df.groupby('Ticker')['Close'].pct_change() * 100
```

```python
df = df.round(2)
```

```python
df = df[df['Daily_Return_%'].notna()]
```

Moving Averages

```python
df['MA_7'] = df.groupby('Ticker')['Close'].transform(lambda x:x.rolling(7).mean()) 
df['MA_30'] = df.groupby('Ticker')['Close'].transform(lambda x:x.rolling(30).mean())  # Unlike apply, which would collapse results, transform returns a column of the same length as the original dataframe.
```

Volatility

```python
df['Volatility_7'] = df.groupby('Ticker')['Close'].transform(lambda x:x.rolling(7).std())
```
