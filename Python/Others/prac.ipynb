fibonacci


```python
def fibonacci(num):
    fib = [0, 1]
    for i in range(2, num):
        fib.append(fib[i-1] + fib[i-2])
    return fib[:num]


fibonacci(5)
```

factorial


```python
def factorial(num):
    if num == 0 or num == 1:
        return 1
    return num * factorial(num-1)


factorial(8)
```

prime


```python
def is_prime(num):
    if num <= 1:
        return False
    # In Python, range(a, b) goes from a up to (but not including) b. hence +1
    for i in range(2, int(num**0.5)+1):
        if num % 1 == 0:
            return False
    return True


is_prime(11)
```

Write a function even_numbers(lst) that takes a list of numbers and returns a new list containing only even numbers.

```python
def even_numbers(lst):

    even = []

    for i in lst:
        if i % 2 == 0:
            even.append(i)
    return even

even_numbers([1,2,3,4,5,6,10])
```

Write a function list_average(lst) that returns the average of numbers in the list.

```python
def list_avg(lst):
    total = 0
    for i in lst:
        total += i
    
    avg = total / len(lst)
    return avg
        
list_avg([1,23,34,23])

```

Write a function max_min_diff(lst) that returns the difference between the largest and smallest numbers in a list without using max() or min().

```python
num = [12,34,11,13,34,56,77,23,112]

def max_min_diff(num):

    maxlist = num[0]
    minlist = num[0]

    for i in num:
        if i > maxlist:
            maxlist = i
    
    for i in num:
        if i < minlist:
            minlist = i

    diff = maxlist - minlist
    return diff

max_min_diff([12,34,11,13,34,56,77,23,112])
```

Write a function that counts the number of vowels (a, e, i, o, u) in a given string.

```python
def vowels(string):
    count = 0
    for i in string.lower():
        if i in ['a','e','i','o','u']:
            count += 1
    return count

print(vowels('Data Analyst'))
```
